## Q09 · Age cohorts

Normalization precedes filtering. Examples are bounded; counts cover the complete source.


```python
# Display Q9 aggregate results from the completed pipeline.
result = dict(report.questions[9])
print(json.dumps(result, ensure_ascii=False, indent=2))
# Draw the two average ages as an inline SVG chart (no plotting dependency).
cohorts = report.questions[9]["summary"]
bars = []
for index, (cohort, values) in enumerate(cohorts.items()):
    age = values["average_age"]
    width = max(0, age or 0) * 3
    y = 40 + index * 70
    label = "No records" if age is None else f"{age:.1f} years; n={values['count']}"
    bars.append(
        f'<text x="8" y="{y}">{cohort}s</text><rect x="85" y="{y - 17}" width="{width}" height="25" fill="#26806b"/><text x="85" y="{y + 28}">{label}</text>'
    )
chart = (
    '<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 900 180" role="img" aria-label="Average age by birth cohort"><rect width="900" height="180" fill="white"/><g fill="#162b3b" font-family="sans-serif">'
    + "".join(bars)
    + "</g></svg>"
)
if "_capture_svg" in globals():
    _capture_svg(chart)  # Export captures the same chart as a notebook output.
else:
    from IPython.display import SVG, display

    display(SVG(chart))
```
